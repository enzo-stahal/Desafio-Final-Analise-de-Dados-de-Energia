```python
df5_solar = df5_solar.rename(columns={'Production': 'Geracao_Solar'})
df5_eolica = df5_eolica.rename(columns={'Production': 'Geracao_Eolica'})
```

```python
max_solar = df5_solar['Geracao_Solar'].max()
max_eolica = df5_eolica['Geracao_Eolica'].max()

print("Máximo de geração solar:", max_solar)
print("Máximo de geração eólica:", max_eolica)
```

```python
# Encontrando os máximos
df5_solar['Geracao_Solar'] = pd.to_numeric(
    df5_solar['Geracao_Solar'], errors='coerce'
)

df5_eolica['Geracao_Eolica'] = pd.to_numeric(
    df5_eolica['Geracao_Eolica'], errors='coerce'
)
```

```python
max_solar = df5_solar['Geracao_Solar'].max()
max_eolica = df5_eolica['Geracao_Eolica'].max()

print("Máximo solar:", max_solar)
print("Máximo eólico:", max_eolica)
```

```python
# Calculando 70% dos máximos
limite_solar = max_solar * 0.70
limite_eolica = max_eolica * 0.70

print("70% do máximo solar:", limite_solar)
print("70% do máximo eólico:", limite_eolica)
```

```python
# Crinado os DatasFrames
limite_solar = max_solar * 0.70
limite_eolica = max_eolica * 0.70

alta_solar = df5_solar[df5_solar['Geracao_Solar'] >= limite_solar].copy()
alta_eolica = df5_eolica[df5_eolica['Geracao_Eolica'] >= limite_eolica].copy()

print("Registros de alta geração solar:", len(alta_solar))
print("Registros de alta geração eólica:", len(alta_eolica))
```

```python
qtd_solar = len(alta_solar)
qtd_eolica = len(alta_eolica)

total_solar = len(df5_solar)
total_eolica = len(df5_eolica)

percentual_solar = (qtd_solar / total_solar) * 100
percentual_eolica = (qtd_eolica / total_eolica) * 100

print(f"Solar: {qtd_solar} registros ({percentual_solar:.2f}%)")
print(f"Eólica: {qtd_eolica} registros ({percentual_eolica:.2f}%)")
```

```python
# Comparando as fontes
if percentual_solar > percentual_eolica:
    print("A geração solar aparece com maior frequência acima de 70% do próprio máximo.")
elif percentual_eolica > percentual_solar:
    print("A geração eólica aparece com maior frequência acima de 70% do próprio máximo.")
else:
    print("As duas fontes apresentam a mesma frequência.")
```

# **Questão 7**

Não é adequado utilizar o mesmo valor numérico como limite para as duas fontes, pois elas possuem escalas de geração diferentes. Neste caso, 70% do máximo solar corresponde a 11.421,20, enquanto 70% do máximo eólico corresponde a 16.050,30. Portanto, cada fonte deve ser comparada com seu próprio máximo.

# **Dataset 6 — Individual Household Electric Power Consumption (UCI)**

**Situação**

Uma residência possui monitoramento elétrico detalhado e deseja identificar episódios de
demanda elevada que também apresentem corrente acima do comportamento médio. O dataset é
o mesmo utilizado como referência em aula, mas o critério de análise será diferente

untitled.svg

```python
#  Renomeando as colunas
df6 = pd.read_csv("/content/household_power_consumption.csv")

df6 = df6.rename(columns={
    "Global_active_power": "Potencia_Ativa",
    "Global_reactive_power": "Potencia_Reativa",
    "Voltage": "Tensao",
    "Global_intensity": "Corrente",
    "Sub_metering_1": "Submetering_1",
    "Sub_metering_2": "Submetering_2",
    "Sub_metering_3": "Submetering_3"
})

df6.head()
```

```python
#  Encontrando a potência ativa máxima
max_potencia = df6["Potencia_Ativa"].max()

print("Potência ativa máxima:", max_potencia)
```

```python
# Calculando 75% do máximo
limite_75_df6 = max_potencia * 0.75

print("Limite de 75%:", limite_75_df6)
```

```python
# Crinado um DataFrame pra 75%
df6_alta_potencia = df6[df6["Potencia_Ativa"] > limite_75_df6]

df6_alta_potencia.head()
```

```python
# Quantidade e percentual dos registros
quantidade_alta = len(df6_alta_potencia)

print("Quantidade de registros:", quantidade_alta)
```

```python
percentual_alta = (quantidade_alta / len(df6)) * 100

print("Percentual de registros:", percentual_alta, "%")
```

```python
# Calculando a corrente média
corrente_media = df6["Corrente"].mean()

print("Corrente média:", corrente_media)
```

```python
# Criando segundo DataFrame
df6_alta_potencia_corrente = df6[
    (df6["Potencia_Ativa"] > limite_75_df6) &
    (df6["Corrente"] > corrente_media)
]

df6_alta_potencia_corrente.head()
```

```python
# Comparando os dois DataFrames
print("Registros com potência acima de 75%:", len(df6_alta_potencia))

print("Registros com potência acima de 75% e corrente acima da média:",
      len(df6_alta_potencia_corrente))
```

```python
percentual_dupla_condicao = (
    len(df6_alta_potencia_corrente) / len(df6)
) * 100

print("Percentual com as duas condições:",
      percentual_dupla_condicao, "%")
```

# **Questão 7**

Ao considerar apenas a potência ativa acima de 75% do máximo, foram selecionados mais registros. Ao adicionar a condição de corrente acima da média, o número de registros diminuiu, pois agora os dados precisam atender simultaneamente às duas condições. Dessa forma, a segunda condição torna a seleção mais restritiva, identificando períodos de alta demanda que também apresentam corrente elevada.
